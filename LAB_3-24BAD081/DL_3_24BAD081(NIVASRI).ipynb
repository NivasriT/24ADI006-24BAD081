## Task A: Dataset Preparation

```python
# NIVASRI T | Roll No: 24BAD081
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Load CIFAR-10 (benchmark dataset, no separate Kaggle download needed)
(x_train_full, y_train_full), (x_test, y_test) = keras.datasets.cifar10.load_data()

class_names = ['airplane', 'automobile', 'bird', 'cat', 'deer',
               'dog', 'frog', 'horse', 'ship', 'truck']

print("Full train shape:", x_train_full.shape)
print("Test shape:", x_test.shape)
```

```python
# NIVASRI T | Roll No: 24BAD081
# Preprocess: normalize pixel values to [0, 1]
x_train_full = x_train_full.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

y_train_full = y_train_full.flatten()
y_test = y_test.flatten()

# Split training data into train and validation sets (90/10)
val_split = int(0.1 * len(x_train_full))
x_val, y_val = x_train_full[:val_split], y_train_full[:val_split]
x_train, y_train = x_train_full[val_split:], y_train_full[val_split:]

print("Train:", x_train.shape, "Validation:", x_val.shape, "Test:", x_test.shape)

# Visualize a few sample images
plt.figure(figsize=(10, 4))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i])
    plt.title(class_names[y_train[i]])
    plt.axis("off")
plt.suptitle("Sample Training Images - NIVASRI T (24BAD081)")
plt.tight_layout()
plt.savefig("task_a_sample_images_24BAD081.png", dpi=150)
plt.show()
```

## Task B: CNN Model Implementation

```python
# NIVASRI T | Roll No: 24BAD081
def build_cnn():
    model = keras.Sequential([
        layers.Input(shape=(32, 32, 3)),

        layers.Conv2D(32, (3, 3), activation="relu", padding="same"),
        layers.Conv2D(32, (3, 3), activation="relu", padding="same"),
        layers.MaxPooling2D((2, 2)),

        layers.Conv2D(64, (3, 3), activation="relu", padding="same"),
        layers.Conv2D(64, (3, 3), activation="relu", padding="same"),
        layers.MaxPooling2D((2, 2)),

        layers.Conv2D(128, (3, 3), activation="relu", padding="same"),
        layers.MaxPooling2D((2, 2)),

        layers.Flatten(),
        layers.Dense(128, activation="relu"),
        layers.Dropout(0.5),
        layers.Dense(10, activation="softmax")
    ])
    return model

model = build_cnn()

model.compile(optimizer="adam",
              loss="sparse_categorical_crossentropy",
              metrics=["accuracy"])

model.summary()
```

## Task C: Model Training and Evaluation

```python
# NIVASRI T | Roll No: 24BAD081
EPOCHS = 15
BATCH_SIZE = 64

history = model.fit(x_train, y_train,
                     validation_data=(x_val, y_val),
                     epochs=EPOCHS,
                     batch_size=BATCH_SIZE,
                     verbose=1)
```

```python
# NIVASRI T | Roll No: 24BAD081
test_loss, test_accuracy = model.evaluate(x_test, y_test, verbose=0)

print(f"Final Training Accuracy:   {history.history['accuracy'][-1]:.4f}")
print(f"Final Validation Accuracy: {history.history['val_accuracy'][-1]:.4f}")
print(f"Final Training Loss:       {history.history['loss'][-1]:.4f}")
print(f"Final Validation Loss:     {history.history['val_loss'][-1]:.4f}")
print(f"Test Accuracy:             {test_accuracy:.4f}")
print(f"Test Loss:                 {test_loss:.4f}")
```

```python
# NIVASRI T | Roll No: 24BAD081
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(history.history["accuracy"], label="Train Accuracy")
axes[0].plot(history.history["val_accuracy"], label="Validation Accuracy")
axes[0].set_title("Accuracy vs Epoch - NIVASRI T (24BAD081)")
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Accuracy")
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].plot(history.history["loss"], label="Train Loss")
axes[1].plot(history.history["val_loss"], label="Validation Loss")
axes[1].set_title("Loss vs Epoch - NIVASRI T (24BAD081)")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Loss")
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig("task_c_training_curves_24BAD081.png", dpi=150)
plt.show()
```

## Task D: Performance Analysis

```python
# NIVASRI T | Roll No: 24BAD081
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns

y_pred_probs = model.predict(x_test)
y_pred = np.argmax(y_pred_probs, axis=1)

# Classification accuracy
print("Overall Test Accuracy:", (y_pred == y_test).mean())
print()
print(classification_report(y_test, y_pred, target_names=class_names))
```

```python
# NIVASRI T | Roll No: 24BAD081
# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(9, 7))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=class_names, yticklabels=class_names)
plt.title("Confusion Matrix - NIVASRI T (24BAD081)")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.tight_layout()
plt.savefig("task_d_confusion_matrix_24BAD081.png", dpi=150)
plt.show()
```

```python
# NIVASRI T | Roll No: 24BAD081
# Sample predictions (correctly classified)
plt.figure(figsize=(12, 5))
correct_idx = np.where(y_pred == y_test)[0][:10]
for i, idx in enumerate(correct_idx):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_test[idx])
    plt.title(f"True: {class_names[y_test[idx]]}\nPred: {class_names[y_pred[idx]]}", fontsize=8)
    plt.axis("off")
plt.suptitle("Sample Correct Predictions - NIVASRI T (24BAD081)")
plt.tight_layout()
plt.savefig("task_d_correct_predictions_24BAD081.png", dpi=150)
plt.show()
```

```python
# NIVASRI T | Roll No: 24BAD081
# Misclassified images
plt.figure(figsize=(12, 5))
misclassified_idx = np.where(y_pred != y_test)[0][:10]
for i, idx in enumerate(misclassified_idx):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_test[idx])
    plt.title(f"True: {class_names[y_test[idx]]}\nPred: {class_names[y_pred[idx]]}", fontsize=8, color="red")
    plt.axis("off")
plt.suptitle("Misclassified Images - NIVASRI T (24BAD081)")
plt.tight_layout()
plt.savefig("task_d_misclassified_24BAD081.png", dpi=150)
plt.show()

print(f"\nTotal misclassified: {(y_pred != y_test).sum()} out of {len(y_test)} test images")
print("\nStrengths: CNNs automatically learn spatial hierarchies of features (edges -> shapes -> objects), "
      "achieve strong accuracy on image data, and need far fewer parameters than a fully-connected network "
      "of comparable depth due to weight sharing in convolution layers.")
print("Limitations: CNNs need large labeled datasets to generalize well, are computationally expensive to "
      "train, can struggle with rotated/occluded objects unless augmented, and confuse visually similar "
      "classes (e.g., cat vs dog, automobile vs truck) as seen in the confusion matrix above.")
```
